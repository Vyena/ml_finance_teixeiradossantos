## Parte 1 — Gradient Boosting desde Cero: Pseudo-Residuos por Función de Pérdida *(25 pts)*

### Contexto

En la sesión 9 vimos el algoritmo de Friedman (2001): dado $F_0(x)$ (una constante), en cada ronda $m$ se calcula el **pseudo-residuo** (el gradiente negativo de la pérdida respecto a las predicciones actuales),

$$r_{im} = -\left[\frac{\partial L(y_i, F(x_i))}{\partial F(x_i)}\right]_{F=F_{m-1}}$$

se ajusta un árbol débil $h_m$ a esos pseudo-residuos, y se actualiza $F_m(x) = F_{m-1}(x) + \nu \cdot h_m(x)$. En clase implementamos esto **solo para pérdida cuadrática** (regresión), donde el pseudo-residuo es simplemente el residuo ordinario $y_i - F_{m-1}(x_i)$. Esta parte te pide completar el algoritmo general: implementarlo para MSE (1.1) y luego derivar y programar el caso de **log-loss** para clasificación binaria (1.2), que es exactamente lo que XGBoost/LightGBM hacen internamente (con una aproximación de segundo orden que aquí no exigimos replicar).

### 1.1 — Gradient Boosting para Regresión (MSE) *(13 pts)*

```python
def gradient_boosting_fit(X, y, n_estimators=100, learning_rate=0.1, max_depth=2, random_state=42):
    '''
    Ajusta un modelo de Gradient Boosting para regresion (perdida MSE) desde cero.

    Sigue el algoritmo de Friedman (2001): en cada ronda, ajusta un arbol a los
    pseudo-residuos (para MSE, el residuo ordinario y - F_(m-1)(x)) y actualiza
    F_m(x) = F_(m-1)(x) + learning_rate * arbol_m(x).

    Parameters
    ----------
    X, y          : arrays de entrenamiento
    n_estimators  : int   — numero de rondas de boosting (M)
    learning_rate : float — shrinkage (nu)
    max_depth     : profundidad de cada arbol debil
    random_state  : semilla maestra (debe generar semillas reproducibles por arbol)

    Returns
    -------
    F0    : float — prediccion constante inicial (media de y)
    trees : list de DecisionTreeRegressor, uno por ronda, ajustados al pseudo-residuo de esa ronda
    '''
    rng = np.random.RandomState(random_state)

    F0 = y.mean()
    F = np.full(len(y), F0, dtype=float)
    trees = []

    for _ in range(n_estimators):
        residual = y - F

        tree = DecisionTreeRegressor(
            max_depth=max_depth,
            random_state=rng.randint(0, 1_000_000)
        )

        tree.fit(X, residual)

        F += learning_rate * tree.predict(X)
        trees.append(tree)

    return F0, trees


def gradient_boosting_predict(F0, trees, X, learning_rate=0.1):
    '''
    Parameters
    ----------
    F0     : constante inicial devuelta por gradient_boosting_fit
    trees  : lista de arboles devuelta por gradient_boosting_fit
    X      : datos sobre los que predecir
    learning_rate : debe ser el mismo usado en el fit

    Returns
    -------
    pred : array — F0 + suma(learning_rate * arbol.predict(X) para cada arbol en trees)
    '''
    pred = np.full(X.shape[0], F0, dtype=float)

    for tree in trees:
        pred += learning_rate * tree.predict(X)

    return pred


# === Verificacion ===

np.random.seed(42)
x_gb = np.linspace(0, 1, 200).reshape(-1, 1)
y_true_gb = np.sin(2 * np.pi * x_gb).ravel()
y_gb = y_true_gb + np.random.normal(0, 0.15, len(x_gb))

N_EST, LR, DEPTH = 100, 0.1, 2
F0_manual, trees_manual = gradient_boosting_fit(x_gb, y_gb, n_estimators=N_EST,
                                                 learning_rate=LR, max_depth=DEPTH, random_state=42)
pred_manual = gradient_boosting_predict(F0_manual, trees_manual, x_gb, learning_rate=LR)

gbr_ref = GradientBoostingRegressor(n_estimators=N_EST, learning_rate=LR, max_depth=DEPTH,
                                     random_state=42).fit(x_gb, y_gb)
pred_ref = gbr_ref.predict(x_gb)

mse_manual = mean_squared_error(y_gb, pred_manual)
mse_ref = mean_squared_error(y_gb, pred_ref)

fig, ax = plt.subplots(figsize=(9, 5))
ax.scatter(x_gb, y_gb, s=10, color="gray", alpha=0.4, label="datos (con ruido)")
ax.plot(x_gb, y_true_gb, "k--", lw=1.5, label="funcion verdadera")
ax.plot(x_gb, pred_manual, lw=2, label=f"Manual (MSE={mse_manual:.4f})")
ax.plot(x_gb, pred_ref, lw=2, ls=":", label=f"sklearn (MSE={mse_ref:.4f})")
ax.legend()
plt.tight_layout()
plt.show()

mse_by_round = []

pred_round = np.full(len(y_gb), F0_manual, dtype=float)

for tree in trees_manual:
    pred_round += LR * tree.predict(x_gb)
    mse_by_round.append(mean_squared_error(y_gb, pred_round))

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(range(1, N_EST + 1), mse_by_round)
ax.set_xlabel("Ronda")
ax.set_ylabel("MSE")
ax.set_title("MSE por ronda — Gradient Boosting manual")
plt.tight_layout()
plt.show()

print(f"MSE manual:  {mse_manual:.4f}")
print(f"MSE sklearn: {mse_ref:.4f}")
print(f"Diferencia relativa: {abs(mse_manual - mse_ref) / mse_ref:.2%}")
```

### 1.2 — Extensión a Clasificación Binaria: Pseudo-Residuo de Log-Loss *(12 pts)*

Para clasificación binaria con log-loss, $F(x)$ representa el **log-odds** (margen, antes de la sigmoide) y la pérdida es $L(y, F) = -[y\log p + (1-y)\log(1-p)]$ con $p = \sigma(F) = 1/(1+e^{-F})$. Deberás derivar (a mano, en la celda de texto siguiente) el pseudo-residuo $r_{im} = -\partial L/\partial F$ para esta pérdida, y usarlo para completar la implementación.

**Requisitos:**
1. En la celda markdown de abajo, muestra el desarrollo de $-\partial L(y,F)/\partial F$ y confirma que se reduce a $y - \sigma(F)$
2. Implementa `gradient_boosting_logloss_fit` y `gradient_boosting_logloss_predict_proba`, reutilizando la misma estructura de 1.1 pero con: (a) $F_0 = \text{logit}(\bar{y})$ (log-odds de la tasa base) en vez de la media, y (b) el pseudo-residuo derivado arriba en vez del residuo ordinario
3. Verifica tu implementación sobre el dataset sintético que genera la celda de verificación (`make_classification`, moderadamente balanceado — a propósito, para no mezclar este chequeo con el problema de desbalance severo de la Parte 3), comparando el AUC de test contra `xgb.XGBClassifier` con los mismos `n_estimators`/`learning_rate`/`max_depth`
4. Tu AUC manual debe quedar dentro de 0.05 del AUC de XGBoost (tu versión es una aproximación de primer orden sin regularización — XGBoost usa una aproximación de segundo orden con $\lambda$, así que una diferencia pequeña es esperable, no un error)

**Derivación (completa aquí):**

$$L(y, F) = -\Big[y\log \sigma(F) + (1-y)\log(1-\sigma(F))\Big], \qquad \sigma(F) = \frac{1}{1+e^{-F}}$$

$$\frac{\partial L}{\partial F} = \; ? \qquad\qquad -\frac{\partial L}{\partial F} = \; ?$$

> Usando que

$$
\sigma'(F)=\sigma(F)(1-\sigma(F)),
$$

derivamos:

$$
\frac{\partial L}{\partial F}
=
-\left[
y\frac{\sigma'(F)}{\sigma(F)}
+
(1-y)\frac{-\sigma'(F)}{1-\sigma(F)}
\right].
$$

Sustituyendo $\sigma'(F)=\sigma(F)(1-\sigma(F))$:

$$
\frac{\partial L}{\partial F}
=
-\left[
y(1-\sigma(F))-(1-y)\sigma(F)
\right]
=
\sigma(F)-y.
$$

Por lo tanto,

$$
-\frac{\partial L}{\partial F}=y-\sigma(F).
$$

Así, el pseudo-residuo en la ronda $m$ es:

$$
r_{im}=y_i-\sigma(F_{m-1}(x_i)).
$$

```python
def gradient_boosting_logloss_fit(X, y, n_estimators=100, learning_rate=0.1, max_depth=2, random_state=42):
    '''
    Ajusta un modelo de Gradient Boosting para clasificacion binaria (log-loss) desde cero.

    F representa el log-odds (margen). El pseudo-residuo de cada ronda es el que
    derivaste arriba: y - sigmoid(F_(m-1)(x)).

    Parameters
    ----------
    X, y          : arrays de entrenamiento (y en {0,1})
    n_estimators  : int
    learning_rate : float
    max_depth     : profundidad de cada arbol debil
    random_state  : semilla maestra

    Returns
    -------
    F0    : float — logit de la tasa base de y (log-odds de y.mean())
    trees : list de DecisionTreeRegressor ajustados al pseudo-residuo logistico de cada ronda
    '''
    rng = np.random.RandomState(random_state)

    p0 = np.clip(y.mean(), 1e-6, 1 - 1e-6)
    F0 = np.log(p0 / (1 - p0))

    F = np.full(len(y), F0, dtype=float)
    trees = []

    for _ in range(n_estimators):
        p = 1 / (1 + np.exp(-F))

        pseudo_residual = y - p

        tree = DecisionTreeRegressor(
            max_depth=max_depth,
            random_state=rng.randint(0, 1_000_000)
        )

        tree.fit(X, pseudo_residual)

        F += learning_rate * tree.predict(X)
        trees.append(tree)

    return F0, trees


def gradient_boosting_logloss_predict_proba(F0, trees, X, learning_rate=0.1):
    '''
    Returns
    -------
    proba : array — sigmoid(F0 + suma(learning_rate * arbol.predict(X)))
    '''
    F = np.full(X.shape[0], F0, dtype=float)

    for tree in trees:
        F += learning_rate * tree.predict(X)

    proba = 1 / (1 + np.exp(-F))

    return proba


# === Verificacion ===

X_synth, y_synth = make_classification(n_samples=2000, n_features=8, n_informative=5,
                                        weights=[0.7, 0.3], class_sep=1.0, random_state=42)
X_synth_tr, X_synth_te, y_synth_tr, y_synth_te = train_test_split(
    X_synth, y_synth, test_size=0.30, stratify=y_synth, random_state=42
)

N_EST2, LR2, DEPTH2 = 100, 0.1, 2
F0_cls, trees_cls = gradient_boosting_logloss_fit(X_synth_tr, y_synth_tr, n_estimators=N_EST2,
                                                   learning_rate=LR2, max_depth=DEPTH2, random_state=42)
proba_manual_cls = gradient_boosting_logloss_predict_proba(F0_cls, trees_cls, X_synth_te, learning_rate=LR2)
auc_manual_cls = roc_auc_score(y_synth_te, proba_manual_cls)

xgb_ref_cls = xgb.XGBClassifier(n_estimators=N_EST2, learning_rate=LR2, max_depth=DEPTH2,
                                 reg_lambda=0.0, eval_metric="auc", random_state=42).fit(X_synth_tr, y_synth_tr)
auc_ref_cls = roc_auc_score(y_synth_te, xgb_ref_cls.predict_proba(X_synth_te)[:, 1])

print(f"AUC manual:  {auc_manual_cls:.4f}")
print(f"AUC XGBoost: {auc_ref_cls:.4f}")
print(f"Diferencia absoluta: {abs(auc_manual_cls - auc_ref_cls):.4f}")
```

> El pseudo-residuo para MSE es simplemente el error actual del modelo: y - F(x).
```python
X_train_full = np.vstack([X_train, X_val])
y_train_full = np.concatenate([y_train, y_val])

# Pipeline + GridSearchCV con TimeSeriesSplit para XGBoost
pipe_xgb = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("clf", xgb.XGBClassifier(eval_metric="auc", random_state=42)),
])

param_grid_xgb = {
    "clf__max_depth": [2, 3, 4],
    "clf__learning_rate": [0.03, 0.1, 0.3],
}

tscv = TimeSeriesSplit(n_splits=5)

grid_xgb = GridSearchCV(
    pipe_xgb,
    param_grid_xgb,
    scoring="roc_auc",
    cv=tscv,
    n_jobs=-1
)

grid_xgb.fit(X_train_full, y_train_full)


# Pipeline + GridSearchCV con TimeSeriesSplit para LightGBM (mismo patron, otro grid)
pipe_lgb = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("clf", lgb.LGBMClassifier(random_state=42, verbosity=-1)),
])

param_grid_lgb = {
    "clf__num_leaves": [7, 15, 31],
    "clf__learning_rate": [0.03, 0.1, 0.3],
}

grid_lgb = GridSearchCV(
    pipe_lgb,
    param_grid_lgb,
    scoring="roc_auc",
    cv=tscv,
    n_jobs=-1
)

grid_lgb.fit(X_train_full, y_train_full)


# Entrenar cada modelo por separado con early stopping sobre X_val, y_val

# XGBoost
xgb_es = xgb.XGBClassifier(
    n_estimators=500,
    learning_rate=0.1,
    max_depth=4,
    eval_metric="auc",
    random_state=42,
    early_stopping_rounds=20
)

xgb_es.fit(
    X_train,
    y_train,
    eval_set=[(X_val, y_val)],
    verbose=False
)


# LightGBM
lgb_es = lgb.LGBMClassifier(
    n_estimators=500,
    learning_rate=0.1,
    num_leaves=15,
    random_state=42,
    verbosity=-1
)

lgb_es.fit(
    X_train,
    y_train,
    eval_set=[(X_val, y_val)],
    callbacks=[lgb.early_stopping(20, verbose=False)]
)


# Tabla comparativa AUC de test — grid search+refit vs early stopping, para XGBoost y LightGBM

# Grid search + refit
proba_xgb_grid = grid_xgb.best_estimator_.predict_proba(X_test)[:, 1]
proba_lgb_grid = grid_lgb.best_estimator_.predict_proba(X_test)[:, 1]

auc_xgb_grid = roc_auc_score(y_test, proba_xgb_grid)
auc_lgb_grid = roc_auc_score(y_test, proba_lgb_grid)

# Early stopping
proba_xgb_es = xgb_es.predict_proba(X_test)[:, 1]
proba_lgb_es = lgb_es.predict_proba(X_test)[:, 1]

auc_xgb_es = roc_auc_score(y_test, proba_xgb_es)
auc_lgb_es = roc_auc_score(y_test, proba_lgb_es)

df_compare = pd.DataFrame({
    "Grid Search + Refit": [auc_xgb_grid, auc_lgb_grid],
    "Early Stopping": [auc_xgb_es, auc_lgb_es]
}, index=["XGBoost", "LightGBM"])

display(df_compare.round(4))

print("Mejores hiperparametros XGBoost:")
print(grid_xgb.best_params_)

print("\nMejores hiperparametros LightGBM:")
print(grid_lgb.best_params_)

print(f"\nBest CV AUC XGBoost: {grid_xgb.best_score_:.4f}")
print(f"Best CV AUC LightGBM: {grid_lgb.best_score_:.4f}")

print(f"\nBest iteration XGBoost ES: {xgb_es.best_iteration}")
print(f"Best iteration LightGBM ES: {lgb_es.best_iteration_}")
```

4. Identifica el feature con **mayor discrepancia** entre su ranking de `Weight` y su ranking de `Gain` (mayor diferencia absoluta de rangos), y en las preguntas de interpretación explica el mecanismo (¿se usa poco pero aporta mucho, o se usa mucho pero aporta poco?)

```python
# Entrenar XGBoost con early stopping (reutiliza X_train/X_val/y_train/y_val)
xgb_imp = xgb.XGBClassifier(
    n_estimators=1000, learning_rate=0.03, max_depth=3,
    eval_metric="auc", early_stopping_rounds=50, random_state=42
)
xgb_imp.fit(X_train, y_train, eval_set=[(X_val, y_val)], verbose=False)

# Extraer weight/gain/cover en un DataFrame indexado por FEATURES
importance_types = ["weight", "gain", "cover"]
booster = xgb_imp.get_booster()

imp_df = pd.DataFrame(index=FEATURES)

for imp_type in importance_types:
    scores = booster.get_score(importance_type=imp_type)
    
    imp_df[imp_type] = [
        scores.get(f"f{i}", 0.0)
        for i in range(len(FEATURES))
    ]

print("Importancias:")
display(imp_df)


# Correlacion de Spearman entre cada par de columnas (3 pares)
corr_spearman = imp_df.corr(method="spearman")

print("\nCorrelaciones de Spearman:")
print(f"Weight-Gain : {corr_spearman.loc['weight', 'gain']:.4f}")
print(f"Weight-Cover: {corr_spearman.loc['weight', 'cover']:.4f}")
print(f"Gain-Cover  : {corr_spearman.loc['gain', 'cover']:.4f}")

# Identificar el feature con mayor |rank_weight - rank_gain|
rank_weight = imp_df["weight"].rank(ascending=False, method="min")
rank_gain = imp_df["gain"].rank(ascending=False, method="min")

imp_df["rank_weight"] = rank_weight
imp_df["rank_gain"] = rank_gain
imp_df["rank_diff"] = (rank_weight - rank_gain).abs()

feature_mayor_discrepancia = imp_df["rank_diff"].idxmax()

print("\nRanking Weight vs Gain:")
display(
    imp_df[
        ["weight", "gain", "cover", "rank_weight", "rank_gain", "rank_diff"]
    ].sort_values("rank_diff", ascending=False)
)

print(f"Feature con mayor discrepancia Weight-Gain: {feature_mayor_discrepancia}")
print(
    f"Rank Weight: {imp_df.loc[feature_mayor_discrepancia, 'rank_weight']:.0f} | "
    f"Rank Gain: {imp_df.loc[feature_mayor_discrepancia, 'rank_gain']:.0f}"
)
```

### 3.2 — Sensibilidad a Ruido de Etiquetas: RF vs XGBoost vs LightGBM, y una Mitigación *(12 pts)*

3. Para cada modelo y nivel de ruido, calcula el **coeficiente de variación** $CV = \text{std}/\text{mean}$ del AUC de test entre semillas — esta es tu métrica cuantitativa de inestabilidad (más alta = más sensible al ruido específico de las etiquetas volteadas)
4. Grafica el CV vs nivel de ruido para los tres modelos en un mismo gráfico

```python
# Experimento de ruido de etiquetas para RF, XGBoost y LightGBM
noise_levels = [0.0, 0.05, 0.10, 0.15, 0.20, 0.30]
n_repeats = 10
noise_results = []

for noise_frac in noise_levels:
    aucs = {"RF": [], "XGBoost": [], "LightGBM": []}

    for rep in range(n_repeats):
        rng = np.random.RandomState(42 + rep)

        y_noisy = y_train.copy()

        n_flip = int(len(y_noisy) * noise_frac)

        if n_flip > 0:
            flip_idx = rng.choice(
                len(y_noisy),
                size=n_flip,
                replace=False
            )
            y_noisy[flip_idx] = 1 - y_noisy[flip_idx]

        rf_noise = RandomForestClassifier(
            n_estimators=200,
            max_features="sqrt",
            min_samples_leaf=10,
            random_state=42 + rep,
            n_jobs=-1
        )

        xgb_noise = xgb.XGBClassifier(
            n_estimators=200,
            learning_rate=0.05,
            max_depth=3,
            eval_metric="auc",
            random_state=42 + rep
        )

        lgb_noise = lgb.LGBMClassifier(
            n_estimators=200,
            learning_rate=0.05,
            num_leaves=15,
            random_state=42 + rep,
            verbosity=-1
        )

        rf_noise.fit(X_train, y_noisy)
        xgb_noise.fit(X_train, y_noisy)
        lgb_noise.fit(X_train, y_noisy)

        aucs["RF"].append(
            roc_auc_score(
                y_test,
                rf_noise.predict_proba(X_test)[:, 1]
            )
        )

        aucs["XGBoost"].append(
            roc_auc_score(
                y_test,
                xgb_noise.predict_proba(X_test)[:, 1]
            )
        )

        aucs["LightGBM"].append(
            roc_auc_score(
                y_test,
                lgb_noise.predict_proba(X_test)[:, 1]
            )
        )

    noise_results.append({
        "noise_frac": noise_frac,

        "RF_mean": np.mean(aucs["RF"]),
        "RF_std": np.std(aucs["RF"]),

        "XGBoost_mean": np.mean(aucs["XGBoost"]),
        "XGBoost_std": np.std(aucs["XGBoost"]),

        "LightGBM_mean": np.mean(aucs["LightGBM"]),
        "LightGBM_std": np.std(aucs["LightGBM"])
    })


noise_df = pd.DataFrame(noise_results)


# Calcular CV = std/mean por modelo y nivel de ruido, graficar vs noise_frac

for model in ["RF", "XGBoost", "LightGBM"]:
    noise_df[f"{model}_CV"] = (
        noise_df[f"{model}_std"] /
        noise_df[f"{model}_mean"]
    )

display(noise_df.round(4))


fig, ax = plt.subplots(figsize=(9, 5))

for model in ["RF", "XGBoost", "LightGBM"]:
    ax.plot(
        noise_df["noise_frac"],
        noise_df[f"{model}_CV"],
        marker="o",
        label=model
    )

ax.set_xlabel("Fraccion de ruido")
ax.set_ylabel("CV del AUC")
ax.set_title("Sensibilidad a ruido de etiquetas")
ax.legend()

plt.tight_layout()
plt.show()


# Mitigacion a noise_frac=0.30 — reentrenar XGBoost/LightGBM con min_child_weight/min_data_in_leaf
# y reg_lambda mas altos, comparar CV y AUC promedio vs la version sin mitigar

noise_frac = 0.30

aucs_xgb_mit = []
aucs_lgb_mit = []

for rep in range(n_repeats):
    rng = np.random.RandomState(42 + rep)

    y_noisy = y_train.copy()

    n_flip = int(len(y_noisy) * noise_frac)

    flip_idx = rng.choice(
        len(y_noisy),
        size=n_flip,
        replace=False
    )

    y_noisy[flip_idx] = 1 - y_noisy[flip_idx]

    xgb_mit = xgb.XGBClassifier(
        n_estimators=200,
        learning_rate=0.05,
        max_depth=3,
        min_child_weight=10,
        reg_lambda=10,
        eval_metric="auc",
        random_state=42 + rep
    )

    lgb_mit = lgb.LGBMClassifier(
        n_estimators=200,
        learning_rate=0.05,
        num_leaves=15,
        min_data_in_leaf=30,
        reg_lambda=10,
        random_state=42 + rep,
        verbosity=-1
    )

    xgb_mit.fit(X_train, y_noisy)
    lgb_mit.fit(X_train, y_noisy)

    aucs_xgb_mit.append(
        roc_auc_score(
            y_test,
            xgb_mit.predict_proba(X_test)[:, 1]
        )
    )

    aucs_lgb_mit.append(
        roc_auc_score(
            y_test,
            lgb_mit.predict_proba(X_test)[:, 1]
        )
    )


base_30 = noise_df.loc[
    noise_df["noise_frac"] == 0.30
].iloc[0]

mitigation_df = pd.DataFrame({
    "Modelo": [
        "XGBoost base",
        "XGBoost mitigado",
        "LightGBM base",
        "LightGBM mitigado"
    ],
    "AUC_mean": [
        base_30["XGBoost_mean"],
        np.mean(aucs_xgb_mit),
        base_30["LightGBM_mean"],
        np.mean(aucs_lgb_mit)
    ],
    "CV": [
        base_30["XGBoost_CV"],
        np.std(aucs_xgb_mit) / np.mean(aucs_xgb_mit),
        base_30["LightGBM_CV"],
        np.std(aucs_lgb_mit) / np.mean(aucs_lgb_mit)
    ]
})

print("\nMitigacion a noise_frac=0.30:")
display(mitigation_df.round(4))
```

```python
# Elegir ventana de crisis (distinta a COVID) y construir rv_20 + lags
CRISIS_START = "2022-01-01"
CRISIS_END   = "2022-10-31"

rv20 = ret.rolling(20).std() * np.sqrt(252)

feat_crisis = pd.DataFrame(index=raw.index)
feat_crisis["rv_lag1"]  = rv20.shift(1)
feat_crisis["rv_lag5"]  = rv20.shift(5)
feat_crisis["rv_lag22"] = rv20.shift(22)
feat_crisis["target_rv"] = rv20
feat_crisis = feat_crisis.dropna()

RV_FEATURES = ["rv_lag1", "rv_lag5", "rv_lag22"]


# Split train (< CRISIS_START) / test (ventana de crisis)
train_mask = feat_crisis.index < CRISIS_START
test_mask = (
    (feat_crisis.index >= CRISIS_START) &
    (feat_crisis.index <= CRISIS_END)
)

X_train_crisis = feat_crisis.loc[train_mask, RV_FEATURES].values
y_train_crisis = feat_crisis.loc[train_mask, "target_rv"].values

X_test_crisis = feat_crisis.loc[test_mask, RV_FEATURES].values
y_test_crisis = feat_crisis.loc[test_mask, "target_rv"].values

dates_crisis = feat_crisis.loc[test_mask].index


# Entrenar XGBoost, LightGBM, RandomForestRegressor, Ridge
xgb_crisis = xgb.XGBRegressor(
    n_estimators=300,
    learning_rate=0.03,
    max_depth=3,
    random_state=42
)

lgb_crisis = lgb.LGBMRegressor(
    n_estimators=300,
    learning_rate=0.03,
    num_leaves=15,
    random_state=42,
    verbosity=-1
)

rf_crisis = RandomForestRegressor(
    n_estimators=300,
    min_samples_leaf=5,
    random_state=42,
    n_jobs=-1
)

ridge_crisis = Ridge(alpha=1.0)

xgb_crisis.fit(X_train_crisis, y_train_crisis)
lgb_crisis.fit(X_train_crisis, y_train_crisis)
rf_crisis.fit(X_train_crisis, y_train_crisis)
ridge_crisis.fit(X_train_crisis, y_train_crisis)

pred_xgb = xgb_crisis.predict(X_test_crisis)
pred_lgb = lgb_crisis.predict(X_test_crisis)
pred_rf = rf_crisis.predict(X_test_crisis)
pred_ridge = ridge_crisis.predict(X_test_crisis)


# Grafico de la serie real vs las 4 predicciones + linea horizontal en max(train)
max_train = np.max(y_train_crisis)

fig, ax = plt.subplots(figsize=(12, 6))

ax.plot(dates_crisis, y_test_crisis, lw=2, label="Real")
ax.plot(dates_crisis, pred_xgb, label="XGBoost")
ax.plot(dates_crisis, pred_lgb, label="LightGBM")
ax.plot(dates_crisis, pred_rf, label="Random Forest")
ax.plot(dates_crisis, pred_ridge, label="Ridge")

ax.axhline(
    max_train,
    linestyle="--",
    label=f"Max train = {max_train:.4f}"
)

ax.set_title(f"Volatilidad realizada — crisis 2022 ({TICKER})")
ax.set_ylabel("RV20")
ax.set_xlabel("Fecha")
ax.legend()
plt.tight_layout()
plt.show()


# Imprimir max(train), max(test), max(pred) para cada modelo, y RMSE de cada uno en la ventana de crisis
print(f"Train: {len(y_train_crisis)} obs (< {CRISIS_START})")
print(f"Test:  {len(y_test_crisis)} obs ({CRISIS_START} a {CRISIS_END})")
print()

print(f"Max y_train       : {np.max(y_train_crisis):.4f}")
print(f"Max y_test        : {np.max(y_test_crisis):.4f}")
print(f"Max pred XGBoost  : {np.max(pred_xgb):.4f}")
print(f"Max pred LightGBM : {np.max(pred_lgb):.4f}")
print(f"Max pred RF       : {np.max(pred_rf):.4f}")
print(f"Max pred Ridge    : {np.max(pred_ridge):.4f}")

print()
print("RMSE en ventana de crisis:")
print(f"XGBoost       : {np.sqrt(mean_squared_error(y_test_crisis, pred_xgb)):.4f}")
print(f"LightGBM      : {np.sqrt(mean_squared_error(y_test_crisis, pred_lgb)):.4f}")
print(f"Random Forest : {np.sqrt(mean_squared_error(y_test_crisis, pred_rf)):.4f}")
print(f"Ridge         : {np.sqrt(mean_squared_error(y_test_crisis, pred_ridge)):.4f}")


# Confirmacion explicita del limite de extrapolacion
print()
print("Limite de extrapolacion:")
print(f"XGBoost       : max pred = {np.max(pred_xgb):.4f} <= max train = {max_train:.4f}")
print(f"LightGBM      : max pred = {np.max(pred_lgb):.4f} <= max train = {max_train:.4f}")
print(f"Random Forest : max pred = {np.max(pred_rf):.4f} <= max train = {max_train:.4f}")
print(f"Ridge         : max pred = {np.max(pred_ridge):.4f}")

print()
if np.max(y_test_crisis) <= max_train:
    print("La ventana seleccionada no exige extrapolacion fuera del rango historico de entrenamiento.")
```

### 4.2 — Costo de una Restricción Monótona Mal Especificada *(12 pts)*

En clase impusimos una restricción monótona **correcta** (`monotone_constraints=(1,)` para PD creciente en DTI). Esta parte te pide medir qué pasa cuando la restricción se especifica **incorrectamente** — un error de dirección plausible si un analista asume la relación equivocada.

**Requisitos:**
1. Reutiliza (o regenera) el escenario sintético PD vs DTI de la sesión 9 (relación creciente verdadera entre DTI y probabilidad de default)
2. Entrena tres versiones de `XGBClassifier` sobre los mismos datos: (a) **sin restricción**, (b) con restricción **correcta** `monotone_constraints=(1,)`, (c) con restricción **incorrecta** `monotone_constraints=(-1,)` (fuerza decreciente cuando la relación real es creciente)
3. Para cada una, calcula: el AUC en un set de test, y si la curva `P(default)` vs `DTI` es monótona (usa `np.diff(pred) >= -1e-9` como en clase)
4. Grafica las tres curvas de predicción vs DTI en un mismo panel
5. Cuantifica el costo: ¿cuánto AUC se pierde con la restricción incorrecta vs sin restricción y vs la correcta?

```python
# Regenerar (o reutilizar) el escenario sintetico PD vs DTI de la sesion 9
np.random.seed(1)
n_mono = 2000
dti_mono = np.random.uniform(0, 1, n_mono)

# relacion verdadera creciente entre DTI y probabilidad de default
logit_mono = -3 + 6 * dti_mono
p_default_mono = 1 / (1 + np.exp(-logit_mono))
default_mono = (np.random.rand(n_mono) < p_default_mono).astype(int)

X_mono = dti_mono.reshape(-1, 1)
X_mono_tr, X_mono_te, y_mono_tr, y_mono_te = train_test_split(
    X_mono, default_mono, test_size=0.3,
    stratify=default_mono, random_state=42
)


# Entrenar 3 versiones — sin restriccion, monotone_constraints=(1,), monotone_constraints=(-1,)
xgb_none = xgb.XGBClassifier(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=3,
    eval_metric="auc",
    random_state=42
)

xgb_correct = xgb.XGBClassifier(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=3,
    monotone_constraints=(1,),
    eval_metric="auc",
    random_state=42
)

xgb_wrong = xgb.XGBClassifier(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=3,
    monotone_constraints=(-1,),
    eval_metric="auc",
    random_state=42
)

xgb_none.fit(X_mono_tr, y_mono_tr)
xgb_correct.fit(X_mono_tr, y_mono_tr)
xgb_wrong.fit(X_mono_tr, y_mono_tr)


# Para cada version, calcular AUC en test y verificar monotonia sobre una grilla de DTI
grid_dti = np.linspace(0, 1, 300).reshape(-1, 1)

models_mono = {
    "Sin restriccion": xgb_none,
    "Restriccion correcta (+1)": xgb_correct,
    "Restriccion incorrecta (-1)": xgb_wrong
}

results_mono = []
pred_grid = {}

for name, model in models_mono.items():

    pred_test = model.predict_proba(X_mono_te)[:, 1]
    auc = roc_auc_score(y_mono_te, pred_test)

    pred = model.predict_proba(grid_dti)[:, 1]
    pred_grid[name] = pred

    creciente = np.all(np.diff(pred) >= -1e-9)
    decreciente = np.all(np.diff(pred) <= 1e-9)

    results_mono.append({
        "Modelo": name,
        "AUC_test": auc,
        "Monotona_creciente": creciente,
        "Monotona_decreciente": decreciente
    })

results_mono_df = pd.DataFrame(results_mono)


# Grafico de las 3 curvas P(default) vs DTI + tabla de AUC por version
fig, ax = plt.subplots(figsize=(9, 5))

for name in models_mono:
    ax.plot(
        grid_dti.ravel(),
        pred_grid[name],
        label=name
    )

ax.set_xlabel("DTI")
ax.set_ylabel("P(default)")
ax.set_title("Restricciones monotonas — P(default) vs DTI")
ax.legend()

plt.tight_layout()
plt.show()

display(results_mono_df.round(4))
```
